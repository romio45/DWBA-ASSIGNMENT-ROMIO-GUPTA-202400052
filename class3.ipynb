```python
!pip install pyspark -q
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum as _sum

spark = SparkSession.builder.appName('RomioSalesCategoryAnalysis').getOrCreate()

sales_records = [
    ('TXN-8801', 'Noise-Cancelling Headphones', 'Consumer Electronics', 14),
    ('TXN-8802', 'Mechanical Keyboard', 'Consumer Electronics', 22),
    ('TXN-8803', 'Ergonomic Desk Chair', 'Office Furniture', 9),
    ('TXN-8804', 'Adjustable Standing Desk', 'Office Furniture', 5),
    ('TXN-8805', 'Ceramic Coffee Mug Set', 'Kitchenware', 30),
    ('TXN-8806', 'Cast Iron Skillet', 'Kitchenware', 18),
    ('TXN-8807', 'Running Shoes', 'Sports & Apparel', 40),
    ('TXN-8808', 'Dri-Fit Workout Shorts', 'Sports & Apparel', 52),
    ('TXN-8809', '4K Ultra HD Monitor', 'Consumer Electronics', 11),
    ('TXN-8810', 'Bookshelf Storage Unit', 'Office Furniture', 7)
]
columns = ['TransactionID', 'ProductTitle', 'Category', 'QuantitySold']
df_sales = spark.createDataFrame(sales_records, schema=columns)
df_sales.groupBy('Category').agg(_sum('QuantitySold').alias('TotalQuantitySold')).orderBy('TotalQuantitySold', ascending=False).show()
```
